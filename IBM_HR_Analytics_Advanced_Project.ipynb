```python

import pandas as pd
import numpy as np

df = pd.read_csv("/content/IBM_HR_Analytics.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python

# Display all column names
print("COLUMN NAMES:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDATASET INFORMATION:")
df.info()
```

```python

# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python

# Count exact duplicate rows
duplicate_count = df.duplicated().sum()

print("Total rows:", len(df))
print("Exact duplicate rows:", duplicate_count)
```

```python

# Check whether EmployeeNumber is unique
duplicate_employee_ids = df["EmployeeNumber"].duplicated().sum()

print("Total employee records:", len(df))
print("Unique employee IDs:", df["EmployeeNumber"].nunique())
print("Duplicate employee IDs:", duplicate_employee_ids)
```

```python

# Check the distinct values in these three columns
columns_to_check = ["EmployeeCount", "Over18", "StandardHours"]

for column in columns_to_check:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False))
```

```python

# Create a separate copy for analysis
df_clean = df.copy()

# Remove columns confirmed to have constant values
columns_to_remove = ["EmployeeCount", "Over18", "StandardHours"]

df_clean = df_clean.drop(columns=columns_to_remove)

print("Original dataset shape:", df.shape)
print("Analysis dataset shape:", df_clean.shape)
print("\nRemaining columns:", df_clean.shape[1])
print("\nRemoved columns:", columns_to_remove)
```

```python

# 1. Check the numerical ranges
print("NUMERICAL SUMMARY:")
display(df_clean.describe().T)

# 2. Check important categorical values
categorical_columns = [
    "Attrition",
    "Department",
    "Gender",
    "OverTime",
    "BusinessTravel",
    "MaritalStatus"
]

print("\nCATEGORICAL VALUE CHECKS:")

for column in categorical_columns:
    print(f"\n{column}:")
    print(df_clean[column].value_counts(dropna=False))

```

```python

rating_columns = [
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "JobInvolvement",
    "WorkLifeBalance"
]

for column in rating_columns:
    print(f"{column}: {sorted(df_clean[column].unique())}")
```

```python

# Create readable labels without changing the original rating columns

satisfaction_labels = {
    1: "Low",
    2: "Medium",
    3: "High",
    4: "Very High"
}

work_life_labels = {
    1: "Bad",
    2: "Good",
    3: "Better",
    4: "Best"
}

df_clean["JobSatisfactionLabel"] = (
    df_clean["JobSatisfaction"].map(satisfaction_labels)
)

df_clean["EnvironmentSatisfactionLabel"] = (
    df_clean["EnvironmentSatisfaction"].map(satisfaction_labels)
)

df_clean["RelationshipSatisfactionLabel"] = (
    df_clean["RelationshipSatisfaction"].map(satisfaction_labels)
)

df_clean["JobInvolvementLabel"] = (
    df_clean["JobInvolvement"].map(satisfaction_labels)
)

df_clean["WorkLifeBalanceLabel"] = (
    df_clean["WorkLifeBalance"].map(work_life_labels)
)

print("Readable rating labels created successfully!")
display(
    df_clean[
        [
            "JobSatisfaction",
            "JobSatisfactionLabel",
            "WorkLifeBalance",
            "WorkLifeBalanceLabel"
        ]
    ].head()
)
```

```python

total_employees = len(df_clean)

employees_left = (df_clean["Attrition"] == "Yes").sum()
employees_stayed = (df_clean["Attrition"] == "No").sum()

attrition_rate = (employees_left / total_employees) * 100
average_age = df_clean["Age"].mean()
average_monthly_income = df_clean["MonthlyIncome"].mean()
average_years_at_company = df_clean["YearsAtCompany"].mean()

print("HR EXECUTIVE METRICS")
print("-" * 35)

print("Total Employees:", total_employees)
print("Employees Who Left:", employees_left)
print("Employees Who Stayed:", employees_stayed)
print(f"Attrition Rate: {attrition_rate:.2f}%")
print(f"Average Age: {average_age:.2f} years")
print(f"Average Monthly Income: {average_monthly_income:.2f}")
print(f"Average Years at Company: {average_years_at_company:.2f}")
```

```python

department_analysis = (
    df_clean.groupby("Department")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

department_analysis["AttritionRate"] = (
    department_analysis["EmployeesLeft"]
    / department_analysis["TotalEmployees"]
    * 100
)

department_analysis = department_analysis.sort_values(
    "AttritionRate", ascending=False
)

department_analysis["AttritionRate"] = (
    department_analysis["AttritionRate"].round(2)
)

display(department_analysis)
```

```python

job_role_analysis = (
    df_clean.groupby("JobRole")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

job_role_analysis["AttritionRate"] = (
    job_role_analysis["EmployeesLeft"]
    / job_role_analysis["TotalEmployees"]
    * 100
)

job_role_analysis = job_role_analysis.sort_values(
    "AttritionRate", ascending=False
)

job_role_analysis["AttritionRate"] = (
    job_role_analysis["AttritionRate"].round(2)
)

display(job_role_analysis)
```

```python

overtime_analysis = (
    df_clean.groupby("OverTime")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

overtime_analysis["AttritionRate"] = (
    overtime_analysis["EmployeesLeft"]
    / overtime_analysis["TotalEmployees"]
    * 100
)

overtime_analysis["AttritionRate"] = (
    overtime_analysis["AttritionRate"].round(2)
)

display(overtime_analysis)
```

```python
income_analysis = (
    df_clean.groupby("Attrition")
    .agg(
        EmployeeCount=("EmployeeNumber", "count"),
        AverageMonthlyIncome=("MonthlyIncome", "mean"),
        MedianMonthlyIncome=("MonthlyIncome", "median")
    )
    .reset_index()
)

income_analysis["AverageMonthlyIncome"] = (
    income_analysis["AverageMonthlyIncome"].round(2)
)

income_analysis["MedianMonthlyIncome"] = (
    income_analysis["MedianMonthlyIncome"].round(2)
)

display(income_analysis)
```

```python
tenure_analysis = (
    df_clean.groupby("Attrition")
    .agg(
        EmployeeCount=("EmployeeNumber", "count"),
        AverageYearsAtCompany=("YearsAtCompany", "mean"),
        MedianYearsAtCompany=("YearsAtCompany", "median")
    )
    .reset_index()
)

tenure_analysis["AverageYearsAtCompany"] = (
    tenure_analysis["AverageYearsAtCompany"].round(2)
)

display(tenure_analysis)
```

```python
satisfaction_analysis = (
    df_clean.groupby("JobSatisfaction")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

satisfaction_analysis["AttritionRate"] = (
    satisfaction_analysis["EmployeesLeft"]
    / satisfaction_analysis["TotalEmployees"]
    * 100
).round(2)

satisfaction_analysis["SatisfactionLabel"] = (
    satisfaction_analysis["JobSatisfaction"].map({
        1: "Low",
        2: "Medium",
        3: "High",
        4: "Very High"
    })
)

display(satisfaction_analysis)
```

```python
travel_analysis = (
    df_clean.groupby("BusinessTravel")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

travel_analysis["AttritionRate"] = (
    travel_analysis["EmployeesLeft"]
    / travel_analysis["TotalEmployees"]
    * 100
).round(2)

display(travel_analysis)
```

```python
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[17, 25, 35, 45, 55, 60],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-60"
    ]
)

age_analysis = (
    df_clean.groupby("AgeGroup", observed=False)
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

age_analysis["AttritionRate"] = (
    age_analysis["EmployeesLeft"]
    / age_analysis["TotalEmployees"]
    * 100
).round(2)

display(age_analysis)
```

```python
gender_analysis = (
    df_clean.groupby("Gender")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

gender_analysis["AttritionRate"] = (
    gender_analysis["EmployeesLeft"]
    / gender_analysis["TotalEmployees"]
    * 100
).round(2)

display(gender_analysis)
```

```python
marital_analysis = (
    df_clean.groupby("MaritalStatus")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

marital_analysis["AttritionRate"] = (
    marital_analysis["EmployeesLeft"]
    / marital_analysis["TotalEmployees"]
    * 100
).round(2)

display(marital_analysis)
```

```python
jobrole_analysis = (
    df_clean.groupby("JobRole")
    .agg(
        TotalEmployees=("EmployeeNumber", "count"),
        EmployeesLeft=("Attrition", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

jobrole_analysis["AttritionRate"] = (
    jobrole_analysis["EmployeesLeft"]
    / jobrole_analysis["TotalEmployees"]
    * 100
).round(2)

jobrole_analysis = jobrole_analysis.sort_values(
    "AttritionRate",
    ascending=False
)

display(jobrole_analysis)
```
