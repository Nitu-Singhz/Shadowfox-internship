```python

# Step 12: Monthly revenue analysis

# Create year-month column
sales_df["YearMonth"] = sales_df["InvoiceDate"].dt.to_period("M")

# Calculate monthly revenue
monthly_revenue = (
    sales_df.groupby("YearMonth", as_index=False)
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Unique_Invoices=("Invoice", "nunique")
    )
)

# Convert YearMonth to text for display and plotting
monthly_revenue["YearMonth"] = monthly_revenue["YearMonth"].astype(str)

# Display monthly summary
display(monthly_revenue)

# Identify highest and lowest revenue months
highest_month = monthly_revenue.loc[
    monthly_revenue["Total_Revenue"].idxmax()
]

lowest_month = monthly_revenue.loc[
    monthly_revenue["Total_Revenue"].idxmin()
]

print("Highest revenue month:", highest_month["YearMonth"])
print("Highest monthly revenue:", round(highest_month["Total_Revenue"], 2))

print("\nLowest revenue month:", lowest_month["YearMonth"])
print("Lowest monthly revenue:", round(lowest_month["Total_Revenue"], 2))

# Plot monthly revenue trend
plt.figure(figsize=(14, 6))
plt.plot(
    monthly_revenue["YearMonth"],
    monthly_revenue["Total_Revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Step 13: Year-over-year monthly revenue comparison

# Convert YearMonth back to a datetime value
monthly_revenue["Month_Date"] = pd.to_datetime(
    monthly_revenue["YearMonth"] + "-01"
)

monthly_revenue["Year"] = monthly_revenue["Month_Date"].dt.year
monthly_revenue["Month"] = monthly_revenue["Month_Date"].dt.month

# Keep months available in both 2010 and 2011
comparison = monthly_revenue[
    monthly_revenue["Year"].isin([2010, 2011])
].pivot(
    index="Month",
    columns="Year",
    values="Total_Revenue"
)

comparison = comparison.dropna(subset=[2010, 2011]).copy()

comparison["Revenue_Change"] = comparison[2011] - comparison[2010]

comparison["YoY_Growth_Percent"] = (
    comparison["Revenue_Change"] / comparison[2010]
) * 100

month_names = {
    1: "January", 2: "February", 3: "March",
    4: "April", 5: "May", 6: "June",
    7: "July", 8: "August", 9: "September",
    10: "October", 11: "November", 12: "December"
}

comparison.insert(
    0,
    "Month_Name",
    comparison.index.map(month_names)
)

display(comparison.round(2))

# Plot year-over-year comparison
plt.figure(figsize=(12, 6))

plt.plot(
    comparison["Month_Name"],
    comparison[2010],
    marker="o",
    label="2010"
)

plt.plot(
    comparison["Month_Name"],
    comparison[2011],
    marker="o",
    label="2011"
)

plt.title("Monthly Revenue Comparison: 2010 vs 2011")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Step 16: Prepare customer data for RFM analysis

# Keep only records with a valid Customer ID
customer_df = sales_df.dropna(subset=["Customer ID"]).copy()

# Set the reference date to one day after the last transaction
reference_date = customer_df["InvoiceDate"].max() + pd.Timedelta(days=1)

# Calculate RFM metrics for each customer
rfm = (
    customer_df.groupby("Customer ID")
    .agg(
        Recency=("InvoiceDate", lambda dates: (reference_date - dates.max()).days),
        Frequency=("Invoice", "nunique"),
        Monetary=("Revenue", "sum")
    )
    .reset_index()
)

# Display the first 10 customers
print("Customers included in RFM analysis:", len(rfm))
print("Reference date:", reference_date)
display(rfm.head(10))

# Summary statistics
print("\nRFM SUMMARY")
display(rfm[["Recency", "Frequency", "Monetary"]].describe().round(2))
```

```python

# Step 17: Calculate RFM scores

# Score Recency: lower number of days is better
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

# Score Frequency: higher invoice count is better
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Score Monetary: higher revenue is better
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Combine scores
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

# Display scored customers
display(
    rfm[
        ["Customer ID", "Recency", "Frequency", "Monetary",
         "R_Score", "F_Score", "M_Score", "RFM_Score"]
    ].head(10)
)

print("\nScore distributions:")
for column in ["R_Score", "F_Score", "M_Score"]:
    print(f"\n{column}")
    print(rfm[column].value_counts().sort_index())
```

```python

# Step 18: Assign customer segments

def assign_segment(row):
    r = row["R_Score"]
    f = row["F_Score"]
    m = row["M_Score"]

    if r >= 3 and f >= 3 and m >= 3:
        return "Champions"
    elif r >= 2 and f >= 3:
        return "Loyal Customers"
    elif r >= 3 and f >= 1 and m >= 2:
        return "Potential Loyalists"
    elif r == 4 and f == 1:
        return "Recent Customers"
    elif r <= 2 and f >= 2 and m >= 2:
        return "At Risk"
    elif r == 1 and f == 1:
        return "Lost Customers"
    else:
        return "Needs Attention"

rfm["Segment"] = rfm.apply(assign_segment, axis=1)

# Count customers in each segment
segment_summary = (
    rfm.groupby("Segment", as_index=False)
    .agg(
        Customer_Count=("Customer ID", "count"),
        Total_Revenue=("Monetary", "sum"),
        Average_Recency=("Recency", "mean"),
        Average_Frequency=("Frequency", "mean"),
        Average_Monetary=("Monetary", "mean")
    )
    .sort_values("Customer_Count", ascending=False)
)

segment_summary["Customer_Percentage"] = (
    segment_summary["Customer_Count"] / len(rfm) * 100
)

display(segment_summary.round(2))

print("Total customers across segments:", segment_summary["Customer_Count"].sum())
print("Total customers in RFM dataset:", len(rfm))
```

```python
# Step 19: Validate RFM customer segments

print("Missing customer segments:", rfm["Segment"].isna().sum())
print("Total customers:", rfm["Customer ID"].nunique())
print("Total segment records:", rfm["Segment"].count())

print("\nCustomer count by segment:")
print(rfm["Segment"].value_counts())

print("\nCheck for duplicate customer IDs:", rfm["Customer ID"].duplicated().sum())
```

```python
# Step 20: Revenue contribution by customer segment

segment_revenue = (
    rfm.groupby("Segment", as_index=False)
    .agg(
        Customer_Count=("Customer ID", "count"),
        Total_Revenue=("Monetary", "sum")
    )
)

total_customer_revenue = segment_revenue["Total_Revenue"].sum()

segment_revenue["Revenue_Percentage"] = (
    segment_revenue["Total_Revenue"] / total_customer_revenue * 100
)

segment_revenue = segment_revenue.sort_values(
    "Total_Revenue", ascending=False
)

display(segment_revenue.round(2))

print("\nTotal customer revenue:", round(total_customer_revenue, 2))
print(
    "Revenue percentage total:",
    round(segment_revenue["Revenue_Percentage"].sum(), 2)
)
```

```python
# Step 21: Customer purchase frequency analysis

purchase_frequency = (
    rfm.groupby("Frequency")
    .agg(Customer_Count=("Customer ID", "count"))
    .reset_index()
    .sort_values("Frequency")
)

purchase_frequency["Customer_Percentage"] = (
    purchase_frequency["Customer_Count"] / len(rfm) * 100
)

one_time_customers = rfm[rfm["Frequency"] == 1].shape[0]
repeat_customers = rfm[rfm["Frequency"] > 1].shape[0]

print("One-time customers:", one_time_customers)
print("Repeat customers:", repeat_customers)
print("One-time customer percentage:", round(one_time_customers / len(rfm) * 100, 2))
print("Repeat customer percentage:", round(repeat_customers / len(rfm) * 100, 2))

print("\nPurchase frequency summary:")
display(purchase_frequency.head(15).round(2))
```

```python
# Step 22: Repeat purchase analysis by customer segment

rfm["Customer_Type"] = np.where(
    rfm["Frequency"] == 1,
    "One-time Customer",
    "Repeat Customer"
)

retention_summary = (
    rfm.groupby(["Segment", "Customer_Type"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

retention_summary["Total_Customers"] = (
    retention_summary["One-time Customer"] +
    retention_summary["Repeat Customer"]
)

retention_summary["Repeat_Customer_Percentage"] = (
    retention_summary["Repeat Customer"] /
    retention_summary["Total_Customers"] * 100
)

retention_summary = retention_summary.sort_values(
    "Repeat_Customer_Percentage",
    ascending=False
)

display(retention_summary.round(2))
```

```python
# Step 23: Overall customer behavior metrics

total_customers = rfm["Customer ID"].nunique()
total_customer_revenue = rfm["Monetary"].sum()
average_customer_revenue = rfm["Monetary"].mean()
median_customer_revenue = rfm["Monetary"].median()
average_invoices = rfm["Frequency"].mean()

one_time_customers = (rfm["Frequency"] == 1).sum()
repeat_customers = (rfm["Frequency"] > 1).sum()

print("Total Customers:", total_customers)
print("Total Customer Revenue:", round(total_customer_revenue, 2))
print("Average Revenue per Customer:", round(average_customer_revenue, 2))
print("Median Revenue per Customer:", round(median_customer_revenue, 2))
print("Average Invoices per Customer:", round(average_invoices, 2))
print("One-time Customers:", one_time_customers)
print("Repeat Customers:", repeat_customers)
print(
    "Repeat Customer Rate (%):",
    round(repeat_customers / total_customers * 100, 2)
)
```

```python
# Step 24: Customer spending by segment

segment_spending = rfm.groupby("Segment").agg(
    Customer_Count=("Customer ID", "nunique"),
    Total_Revenue=("Monetary", "sum"),
    Average_Revenue=("Monetary", "mean"),
    Median_Revenue=("Monetary", "median")
).reset_index()

segment_spending = segment_spending.sort_values(
    by="Total_Revenue", ascending=False
)

segment_spending[[
    "Total_Revenue", "Average_Revenue", "Median_Revenue"
]] = segment_spending[[
    "Total_Revenue", "Average_Revenue", "Median_Revenue"
]].round(2)

print(segment_spending.to_string(index=False))
```

```python
# Step 25: Revenue contribution by customer segment

segment_spending["Revenue_Percentage"] = (
    segment_spending["Total_Revenue"]
    / segment_spending["Total_Revenue"].sum()
    * 100
).round(2)

segment_spending = segment_spending.sort_values(
    by="Revenue_Percentage", ascending=False
)

print(segment_spending[
    ["Segment", "Customer_Count", "Total_Revenue", "Revenue_Percentage"]
].to_string(index=False))
```

```python
# Step 26: Top 10 Champions by customer revenue

top_champions = rfm[rfm["Segment"] == "Champions"].sort_values(
    by="Monetary", ascending=False
).head(10)

print(top_champions[
    ["Customer ID", "Recency", "Frequency", "Monetary", "RFM_Score"]
].round(2).to_string(index=False))
```

```python
# Step 27: Revenue comparison — one-time vs repeat customers

customer_type_analysis = rfm.groupby("Customer_Type").agg(
    Customer_Count=("Customer ID", "nunique"),
    Total_Revenue=("Monetary", "sum"),
    Average_Revenue=("Monetary", "mean"),
    Average_Invoices=("Frequency", "mean")
).reset_index()

customer_type_analysis["Revenue_Percentage"] = (
    customer_type_analysis["Total_Revenue"]
    / customer_type_analysis["Total_Revenue"].sum()
    * 100
).round(2)

customer_type_analysis[
    ["Total_Revenue", "Average_Revenue", "Average_Invoices"]
] = customer_type_analysis[
    ["Total_Revenue", "Average_Revenue", "Average_Invoices"]
].round(2)

print(customer_type_analysis.to_string(index=False))
```

```python
# Step 34: Average Order Value by Country

country_aov = customer_df.groupby("Country").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("Invoice", "nunique")
).reset_index()

country_aov["Average_Order_Value"] = (
    country_aov["Total_Revenue"] / country_aov["Total_Orders"]
).round(2)

country_aov = country_aov.sort_values(
    "Average_Order_Value", ascending=False
)

print(country_aov.head(10).to_string(index=False))
```

```python
# Step 35: Countries with the most orders

country_order_volume = country_aov.sort_values(
    "Total_Orders", ascending=False
)

print(
    country_order_volume[
        ["Country", "Total_Orders", "Total_Revenue", "Average_Order_Value"]
    ].head(10).to_string(index=False)
)
```

```python
# Step 36: Yearly sales performance

customer_df["Year"] = customer_df["InvoiceDate"].dt.year

yearly_analysis = customer_df.groupby("Year").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("Invoice", "nunique"),
    Unique_Customers=("Customer ID", "nunique")
).reset_index()

yearly_analysis["Average_Order_Value"] = (
    yearly_analysis["Total_Revenue"] /
    yearly_analysis["Total_Orders"]
).round(2)

yearly_analysis["Total_Revenue"] = (
    yearly_analysis["Total_Revenue"].round(2)
)

print(yearly_analysis.to_string(index=False))
```

```python
# Step 37: Yearly sales by product

sales_df["Year"] = sales_df["InvoiceDate"].dt.year

yearly_product_sales = sales_df.groupby(
    ["Year", "StockCode", "Description"]
).agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("Invoice", "nunique")
).reset_index()

yearly_product_sales = yearly_product_sales.sort_values(
    ["Year", "Total_Revenue"],
    ascending=[True, False]
)

top_products_each_year = (
    yearly_product_sales.groupby("Year").head(5)
)

print(top_products_each_year.to_string(index=False))
```

```python
# Step 38: Top 10 products by total quantity sold

product_quantity_analysis = sales_df.groupby(
    ["StockCode", "Description"]
).agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("Invoice", "nunique")
).reset_index()

top_10_by_quantity = product_quantity_analysis.sort_values(
    "Total_Quantity", ascending=False
).head(10)

print(top_10_by_quantity.to_string(index=False))
```

```python
# Step 39: Top 10 products by revenue contribution

product_quantity_analysis["Revenue_Percentage"] = (
    product_quantity_analysis["Total_Revenue"]
    / sales_df["Revenue"].sum()
    * 100
).round(2)

top_10_by_revenue = product_quantity_analysis.sort_values(
    "Total_Revenue", ascending=False
).head(10)

print(
    top_10_by_revenue[
        ["StockCode", "Description", "Total_Revenue",
         "Total_Quantity", "Total_Orders", "Revenue_Percentage"]
    ].to_string(index=False)
)
```
